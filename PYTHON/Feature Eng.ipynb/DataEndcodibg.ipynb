### label Encoding
#### --This is used to encode categorical data as numerical data,they are usually assigned in alphabetical order or base on the frequency of the category

```python
df = pd.DataFrame({
    'color':['red','blue','yellow','green','white']
})
df.head()
```

```python
from sklearn.preprocessing import LabelEncoder
lbl_encoder = LabelEncoder()
```

```python
lbl_encoder.fit_transform(df[['color']])
```

```python
lbl_encoder.transform([['blue']])
```

```python
lbl_encoder.transform([['yellow']])
```

```python
lbl_encoder.transform([['green']])
```

#### ORDINAL ENCODING
##### This is used to encode categorical data that have an intimistic code or ranking,each category is assigned to a numerical value based on its position in the order

```python
## Ordinal encoder 

from sklearn.preprocessing import OrdinalEncoder
ordinal_encoder = OrdinalEncoder()
###
```

```python
#### create a simple dataframe with an orinal variable
df=pd.DataFrame({
    'size':['small','medium','large','extralarge','largest']
})
```

##### TARGET GUIDED ORDINAL ENCODING
##### ---This is used to encode categorical variables based on their relationship qith the target variable,mean and median is also used and it usually deals with large data

```python
import pandas as pd
df=pd.DataFrame({
    'city':['lagos','Abuja','kano','Kaduna','Benin'],
    'price':[200,300,400,500,600]
})
```

```python
df
```

```python
df.groupby('city')['price'].median()
```

```python
df.groupby('city')['price'].median().to_dict()
```

```python
mean_price=df.groupby('city')['price'].mean().to_dict
```

```python
mean_price
```

```python
df['city_encoded']=df['city'].map(mean_price)
```
